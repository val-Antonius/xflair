```python
import pandas as pd

df = pd.read_csv('data/processed/wc2022sb_dribbles_clean.csv', low_memory=False)
print(df.shape)
```

```python
def get_player_info(df, player_id, max_rows=50):
    """Return player summary and related rows for a given player_id."""
    if 'player_id' not in df.columns:
        raise KeyError("Column 'player_id' not found. Check df.columns for the correct field name.")
    if player_id is None:
        raise ValueError("player_id must be provided")

    player_rows = df[df['player_id'] == player_id]
    if player_rows.empty:
        return {
            'player_id': player_id,
            'rows_found': 0,
            'message': 'No rows found for this player_id.'
        }

    # Capture common identity fields if they exist
    identity_fields = [col for col in ['player_name', 'player', 'player_id'] if col in player_rows.columns]
    identity = player_rows[identity_fields].drop_duplicates().to_dict('records')

    # Provide a small sample of related rows
    sample = player_rows.head(max_rows).copy()

    return {
        'player_id': player_id,
        'rows_found': int(player_rows.shape[0]),
        'identity': identity,
        'sample': sample
    }

# Example usage
result = get_player_info(df, 25104, max_rows=10)
print('rows_found:', result['rows_found'])
print('identity:', result.get('identity'))
result['sample'].head()
```
