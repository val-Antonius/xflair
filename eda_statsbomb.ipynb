```python
# List competitions with country name and gender
competitions = sb.competitions()
cols = ['competition_name', 'country_name', 'competition_gender']
missing = [c for c in cols if c not in competitions.columns]
if missing:
    raise KeyError(f"Missing columns in competitions: {missing}")

comp_view = (
    competitions[cols]
    .dropna(subset=['competition_name'])
    .drop_duplicates()
    .sort_values(['competition_name', 'country_name', 'competition_gender'])
    .reset_index(drop=True)
 )

for i, row in comp_view.iterrows():
    print(f"{i+1:02d}. {row['competition_name']} | {row['country_name']} | {row['competition_gender']}")
```

```python
# Filter out female competitions and specified regions
exclude_countries = {'europe', 'south america', 'international', 'africa'}
filtered = comp_view[
    (comp_view['competition_gender'].str.lower() != 'female')
    & (~comp_view['country_name'].str.lower().isin(exclude_countries))
].reset_index(drop=True)

for i, row in filtered.iterrows():
    print(f"{i+1:02d}. {row['competition_name']} | {row['country_name']} | {row['competition_gender']}")
```

```python
import pandas as pd
from statsbombpy import sb

def download_events(competition_id, season_id, competition_name):
    matches = sb.matches(competition_id=competition_id, season_id=season_id)
    all_events = []
    for match_id in matches['match_id'].tolist():
        events = sb.events(match_id=match_id)
        all_events.append(events)
    df = pd.concat(all_events, ignore_index=True)
    print(f"{competition_name}: {len(df)} events")
    return df

euro_events = download_events(55, 282, "UEFA Euro 2024")
copa_events = download_events(223, 282, "Copa America 2024")
```

```python
# read WC 2022 events
wc_events = pd.read_csv('data/raw/wc2022_events.csv', low_memory=False)

# add label kompetisi
wc_events['competition'] = 'WC 2022'
euro_events['competition'] = 'Euro 2024'
copa_events['competition'] = 'Copa America 2024'

# mix 3 events
all_events = pd.concat([wc_events, euro_events, copa_events], ignore_index=True)

print(f"WC 2022: {len(wc_events)} events")
print(f"Euro 2024: {len(euro_events)} events")
print(f"Copa America 2024: {len(copa_events)} events")
print(f"Total gabungan: {len(all_events)} events")
```

```python
all_events.to_csv('data/raw/all_events_combined.csv', index=False)
print("Tersimpan!")
```
